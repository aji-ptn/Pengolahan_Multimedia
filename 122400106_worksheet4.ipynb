### Soal 1 — Cropping dan Konversi Warna
Pada soal 1 saya menggunakan gambar `selfie.jpeg` untuk melakukan cropping wajah dan latar belakang. Hasil crop kemudian diubah ukuran menjadi 920×920 piksel, dikonversi ke grayscale dan HSV, serta diberi teks nama di atas kepala.


````python
import cv2
import matplotlib.pyplot as plt
from pathlib import Path

assets = Path("SahatGultom_1220400106/assets_ws4")
results = Path("SahatGultom_1220400106/results_ws4")
results.mkdir(parents=True, exist_ok=True)

img = cv2.imread(str(assets / "selfie.jpeg"))
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

face_crop = img_rgb[200:900, 300:1000]  # crop area wajah
background_crop = img_rgb[100:900, 1100:1800]  # crop area latar belakang

face_resize = cv2.resize(face_crop, (920, 920))
background_resize = cv2.resize(background_crop, (920, 920))

img_annotated = face_resize.copy()
cv2.putText(img_annotated, "SAHAT JUNVIER DANNY TREDO GULTOM", (30, 80), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 0, 0), 2, cv2.LINE_AA)

img_gray = cv2.cvtColor(img_annotated, cv2.COLOR_RGB2GRAY)
img_hsv = cv2.cvtColor(img_annotated, cv2.COLOR_RGB2HSV)
img_hsv_display = cv2.cvtColor(img_hsv, cv2.COLOR_HSV2RGB)

fig, ax = plt.subplots(1, 3, figsize=(18, 6))
ax[0].imshow(img_annotated)
ax[0].set_title("Hasil Crop + Anotasi")
ax[0].axis("off")
ax[1].imshow(img_gray, cmap="gray")
ax[1].set_title("Grayscale")
ax[1].axis("off")
ax[2].imshow(img_hsv_display)
ax[2].set_title("HSV (ditampilkan pada RGB)")
ax[2].axis("off")
plt.show()

plt.figure(figsize=(8, 8))
plt.imshow(background_resize)
plt.title("Crop Latar Belakang")
plt.axis("off")
plt.show()

cv2.imwrite(str(results / "soal1_crop_face_annotated.png"), cv2.cvtColor(img_annotated, cv2.COLOR_RGB2BGR))
cv2.imwrite(str(results / "soal1_crop_background.png"), cv2.cvtColor(background_resize, cv2.COLOR_RGB2BGR))
````

**Penjelasan Soal 1:**
Cropping memusatkan perhatian pada bagian wajah untuk menonjolkan objek utama, sedangkan crop latar belakang menampilkan konteks lingkungan. Konversi ke grayscale menonjolkan perbedaan intensitas tanpa warna, sedangkan representasi HSV membantu memahami komponen warna secara terpisah. Anotasi nama memperjelas identitas hasil.


### Soal 2 — Manipulasi Channel Warna RGB
Soal 2 menggunakan crop wajah dari Soal 1 sebagai input. Channel merah ditingkatkan dan channel biru dikurangi untuk melihat perubahan tonal warna yang terjadi.


````python
import numpy as np

img_crop = face_resize.copy()
R = img_crop[:, :, 0].astype(np.int32)
G = img_crop[:, :, 1].astype(np.int32)
B = img_crop[:, :, 2].astype(np.int32)

R = np.clip(R + 50, 0, 255)
G = np.clip(G, 0, 255)
B = np.clip(B - 30, 0, 255)

img_modified = np.stack([R, G, B], axis=2).astype(np.uint8)

cv2.imwrite(str(results / "soal2_rgb_modified.png"), cv2.cvtColor(img_modified, cv2.COLOR_RGB2BGR))

fig, ax = plt.subplots(1, 2, figsize=(14, 6))
ax[0].imshow(img_crop)
ax[0].set_title("Original Crop Wajah")
ax[0].axis("off")
ax[1].imshow(img_modified)
ax[1].set_title("Hasil Modifikasi RGB")
ax[1].axis("off")
plt.show()

plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.hist(img_crop[:, :, 0].ravel(), bins=256, color="red", alpha=0.5, label="R")
plt.hist(img_crop[:, :, 1].ravel(), bins=256, color="green", alpha=0.5, label="G")
plt.hist(img_crop[:, :, 2].ravel(), bins=256, color="blue", alpha=0.5, label="B")
plt.title("Histogram RGB Asli")
plt.legend()

plt.subplot(1, 2, 2)
plt.hist(img_modified[:, :, 0].ravel(), bins=256, color="red", alpha=0.5, label="R")
plt.hist(img_modified[:, :, 1].ravel(), bins=256, color="green", alpha=0.5, label="G")
plt.hist(img_modified[:, :, 2].ravel(), bins=256, color="blue", alpha=0.5, label="B")
plt.title("Histogram RGB Modifikasi")
plt.legend()
plt.tight_layout()
plt.show()
````

**Penjelasan Soal 2:**
Peningkatan channel merah dan penurunan channel biru membuat gambar terlihat lebih hangat dan menonjolkan warna kulit. Diagram histogram menunjukkan pergeseran distribusi nilai R ke kanan dan nilai B ke kiri.


### Soal 3 — Deteksi Tepi dan Filter Citra
Soal 3 menggunakan gambar `jetson.jpg` untuk mendeteksi tepi dan menerapkan thresholding, bounding box, blur, dan sharpening.


````python
img_obj = cv2.imread(str(assets / "jetson.jpg"))
img_obj_rgb = cv2.cvtColor(img_obj, cv2.COLOR_BGR2RGB)
gray_obj = cv2.cvtColor(img_obj, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray_obj, 80, 180)
_, thresh = cv2.threshold(gray_obj, 120, 255, cv2.THRESH_BINARY)

contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
img_bbox = img_obj_rgb.copy()
for c in contours:
    area = cv2.contourArea(c)
    if area > 1000:
        x, y, w, h = cv2.boundingRect(c)
        cv2.rectangle(img_bbox, (x, y), (x + w, y + h), (255, 0, 0), 3)

img_blur = cv2.GaussianBlur(img_obj_rgb, (15, 15), 0)
kernel_sharp = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
img_sharp = cv2.filter2D(img_obj_rgb, -1, kernel_sharp)

cv2.imwrite(str(results / "soal3_edges.png"), cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR))
cv2.imwrite(str(results / "soal3_threshold.png"), cv2.cvtColor(thresh, cv2.COLOR_GRAY2BGR))
cv2.imwrite(str(results / "soal3_bbox.png"), cv2.cvtColor(img_bbox, cv2.COLOR_RGB2BGR))
cv2.imwrite(str(results / "soal3_blur.png"), cv2.cvtColor(img_blur, cv2.COLOR_RGB2BGR))
cv2.imwrite(str(results / "soal3_sharp.png"), cv2.cvtColor(img_sharp, cv2.COLOR_RGB2BGR))

fig, ax = plt.subplots(2, 3, figsize=(18, 10))
ax[0, 0].imshow(img_obj_rgb)
ax[0, 0].set_title("Original")
ax[0, 0].axis("off")
ax[0, 1].imshow(edges, cmap="gray")
ax[0, 1].set_title("Canny Edge")
ax[0, 1].axis("off")
ax[0, 2].imshow(thresh, cmap="gray")
ax[0, 2].set_title("Threshold")
ax[0, 2].axis("off")
ax[1, 0].imshow(img_bbox)
ax[1, 0].set_title("Bounding Box")
ax[1, 0].axis("off")
ax[1, 1].imshow(img_blur)
ax[1, 1].set_title("Blur Filter")
ax[1, 1].axis("off")
ax[1, 2].imshow(img_sharp)
ax[1, 2].set_title("Sharpen Filter")
ax[1, 2].axis("off")
plt.show()
````

**Penjelasan Soal 3:**
Deteksi tepi mengekspos batas objek pada latar belakang bertekstur, sementara thresholding membantu mencari area dominan. Zoom pada bounding box menunjukkan objek utama yang tersaring. Blur menurunkan detail halus, sedangkan sharpen meningkatkan kontras detail.


### Soal 4 — Deteksi Wajah dan Filter Digital Kreatif
Soal 4 menggunakan MediaPipe Face Mesh untuk mendeteksi landmark wajah dan menambahkan overlay filter kreatif pada gambar `reddevil.jpg`.


````python
import mediapipe as mp

img_face = cv2.imread(str(assets / "reddevil.jpg"))
img_face_rgb = cv2.cvtColor(img_face, cv2.COLOR_BGR2RGB)

mp_face_mesh = mp.solutions.face_mesh
face_mesh = mp_face_mesh.FaceMesh(static_image_mode=True, max_num_faces=1)
results_face = face_mesh.process(img_face_rgb)

img_filter = img_face_rgb.copy()

if results_face.multi_face_landmarks:
    h, w, _ = img_face_rgb.shape
    for face in results_face.multi_face_landmarks:
        left_eye = face.landmark[33]
        right_eye = face.landmark[263]
        nose = face.landmark[1]
        lx, ly = int(left_eye.x * w), int(left_eye.y * h)
        rx, ry = int(right_eye.x * w), int(right_eye.y * h)
        nx, ny = int(nose.x * w), int(nose.y * h)
        cv2.circle(img_filter, (lx, ly), 40, (0, 255, 0), 4)
        cv2.circle(img_filter, (rx, ry), 40, (0, 255, 0), 4)
        cv2.line(img_filter, (lx + 40, ly), (rx - 40, ry), (0, 255, 0), 4)
        overlay = img_filter.copy()
        cv2.rectangle(overlay, (nx - 60, ny + 20), (nx + 60, ny + 80), (255, 0, 0), -1)
        alpha = 0.6
        cv2.addWeighted(overlay, alpha, img_filter, 1 - alpha, 0, img_filter)
        cv2.putText(img_filter, "FILTER CREATIVE", (nx - 100, ny - 30), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2, cv2.LINE_AA)

cv2.imwrite(str(results / "soal4_filter.png"), cv2.cvtColor(img_filter, cv2.COLOR_RGB2BGR))

fig, ax = plt.subplots(1, 2, figsize=(16, 8))
ax[0].imshow(img_face_rgb)
ax[0].set_title("Original")
ax[0].axis("off")
ax[1].imshow(img_filter)
ax[1].set_title("Hasil Filter Digital")
ax[1].axis("off")
plt.show()
````

**Penjelasan Soal 4:**
Overlay ditempatkan berdasarkan landmark wajah: posisi mata dan hidung dijadikan acuan. Alpha blending membantu membuat filter tampak lebih natural dan menyatu dengan wajah.


### Soal 5 — Perspektif dan Peningkatan Kualitas Citra
Soal 5 menampilkan proses koreksi perspektif dan thresholding Otsu pada gambar `kertas.jpg` sehingga teks dan dokumen lebih tajam.


````python
img_doc = cv2.imread(str(assets / "kertas.jpg"))
img_doc_rgb = cv2.cvtColor(img_doc, cv2.COLOR_BGR2RGB)
gray_doc = cv2.cvtColor(img_doc, cv2.COLOR_BGR2GRAY)

pts_src = np.float32([[120, 100], [850, 120], [80, 700], [900, 720]])
width, height = 800, 600
pts_dst = np.float32([[0, 0], [width, 0], [0, height], [width, height]])
matrix = cv2.getPerspectiveTransform(pts_src, pts_dst)
img_warp = cv2.warpPerspective(img_doc_rgb, matrix, (width, height))
warp_gray = cv2.cvtColor(img_warp, cv2.COLOR_RGB2GRAY)
_, otsu = cv2.threshold(warp_gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

cv2.imwrite(str(results / "soal5_perspective.png"), cv2.cvtColor(img_warp, cv2.COLOR_RGB2BGR))
cv2.imwrite(str(results / "soal5_otsu.png"), otsu)

fig, ax = plt.subplots(2, 2, figsize=(16, 12))
ax[0, 0].imshow(img_doc_rgb)
ax[0, 0].set_title("Original")
ax[0, 0].axis("off")
ax[0, 1].imshow(gray_doc, cmap="gray")
ax[0, 1].set_title("Grayscale")
ax[0, 1].axis("off")
ax[1, 0].imshow(img_warp)
ax[1, 0].set_title("Koreksi Perspektif")
ax[1, 0].axis("off")
ax[1, 1].imshow(otsu, cmap="gray")
ax[1, 1].set_title("Otsu Thresholding")
ax[1, 1].axis("off")
plt.show()
````

**Penjelasan Soal 5:**
Koreksi perspektif memperbaiki distorsi sudut sehingga objek datar tampil sejajar. Thresholding Otsu dipilih karena secara otomatis menentukan nilai ambang ideal berdasarkan histogram, menjadikan objek lebih jelas tanpa pengaturan manual.
