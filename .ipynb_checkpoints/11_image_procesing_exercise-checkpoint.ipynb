### **Soal 2 — Manipulasi Channel Warna RGB**

* Gunakan gambar hasil crop dari Soal 1.
* Konversikan gambar ke ruang warna **RGB**.
* Lakukan manipulasi channel warna dengan cara:

  * **Naikkan intensitas channel merah sebanyak 50 poin** (maksimum 255).
  * **Turunkan intensitas channel biru sebanyak 30 poin** (minimum 0).
* Teknik atau cara menaikkan/menurunkan intensitas **dibebaskan**, asalkan logis dan hasilnya terlihat.
* Gabungkan kembali channel warna dan **simpan gambar hasil modifikasi dalam format `.png`**.
* **Tampilkan histogram per channel (R, G, B)** untuk gambar asli dan hasil modifikasi menggunakan `matplotlib.pyplot.hist`.
* Jelaskan dampak perubahan RGB pada warna gambar dalam sel **Markdown**.


````python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ── Path file hasil crop dari Soal 1 ──────────
OUTPUT_DIR = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\output"
import os

# ── Load crop wajah (dengan fallback ke variabel jika ada) ──
try:
    img_bgr = crop_wajah_annotated.copy()
    print("✔ Menggunakan variabel crop_wajah_annotated dari Soal 1")
except NameError:
    path_wajah = os.path.join(OUTPUT_DIR, "crop_wajah_annotated.png")
    img_bgr = cv2.imread(path_wajah)
    if img_bgr is None:
        raise FileNotFoundError(f"File tidak ditemukan: {path_wajah}\n"
                                 "Jalankan Soal 1 terlebih dahulu!")
    print(f"✔ Gambar dimuat dari file: {path_wajah}")

# Lanjut seperti biasa...
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
R, G, B = img_rgb[:,:,0], img_rgb[:,:,1], img_rgb[:,:,2]

# ── Manipulasi channel ────────────────────────
# Naikkan Red +50, turunkan Blue -30 (clip agar tetap 0–255)
R_baru = np.clip(R.astype(np.int16) + 50, 0, 255).astype(np.uint8)
B_baru = np.clip(B.astype(np.int16) - 30, 0, 255).astype(np.uint8)
G_baru = G.copy()  # Green tidak diubah

# ── Gabungkan kembali channel ─────────────────
img_modif = np.stack([R_baru, G_baru, B_baru], axis=2)

# ── Simpan hasil ke folder output ────────────
import os
OUTPUT_DIR = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

img_modif_bgr = cv2.cvtColor(img_modif, cv2.COLOR_RGB2BGR)
path_hasil = os.path.join(OUTPUT_DIR, "hasil_modifikasi_rgb.png")
cv2.imwrite(path_hasil, img_modif_bgr)
print(f"✔ Gambar tersimpan: {path_hasil}")

# ─────────────────────────────────────────────
# VISUALISASI — Gambar berdampingan
# ─────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.patch.set_facecolor('#1a1a2e')

for ax, img, judul in zip(axes,
                           [img_rgb, img_modif],
                           ["Original RGB", "Modifikasi (R+50, B-30)"]):
    ax.imshow(img)
    ax.set_title(judul, fontsize=13, fontweight='bold',
                 color='white', fontfamily='monospace')
    ax.axis('off')

plt.suptitle("Soal 2 — Perbandingan Gambar", fontsize=15,
             fontweight='bold', color='white', fontfamily='monospace')
plt.tight_layout()
plt.show()

# ─────────────────────────────────────────────
# HISTOGRAM per channel — Asli vs Modifikasi
# ─────────────────────────────────────────────
nama_channel   = ['Red', 'Green', 'Blue']
channels_ori   = [img_rgb[:,:,0],   img_rgb[:,:,1],   img_rgb[:,:,2]]
channels_modif = [img_modif[:,:,0], img_modif[:,:,1], img_modif[:,:,2]]
warna_plot     = ['#ff4444', '#44ff88', '#4488ff']

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
fig.patch.set_facecolor('#1a1a2e')

for col, (ch_ori, ch_mod, warna, nama_ch) in enumerate(
        zip(channels_ori, channels_modif, warna_plot, nama_channel)):

    # Baris 0 — histogram asli
    axes[0, col].hist(ch_ori.ravel(), bins=256, range=(0,255),
                      color=warna, alpha=0.85, edgecolor='none')
    axes[0, col].set_title(f'{nama_ch} — Original',
                           color='white', fontsize=11, fontweight='bold')
    axes[0, col].set_facecolor('#0d0d1a')
    axes[0, col].tick_params(colors='white')
    axes[0, col].set_xlim(0, 255)

    # Baris 1 — histogram modifikasi
    axes[1, col].hist(ch_mod.ravel(), bins=256, range=(0,255),
                      color=warna, alpha=0.85, edgecolor='none')
    axes[1, col].set_title(f'{nama_ch} — Modifikasi',
                           color='white', fontsize=11, fontweight='bold')
    axes[1, col].set_facecolor('#0d0d1a')
    axes[1, col].tick_params(colors='white')
    axes[1, col].set_xlim(0, 255)

    # Label sumbu
    for row in range(2):
        axes[row, col].set_xlabel("Intensitas (0–255)", color='#aaaaaa', fontsize=9)
        axes[row, col].set_ylabel("Jumlah Piksel",      color='#aaaaaa', fontsize=9)
        for spine in axes[row, col].spines.values():
            spine.set_edgecolor('#333355')

plt.suptitle("Soal 2 — Histogram per Channel (Asli vs Modifikasi)",
             fontsize=14, fontweight='bold', color='white',
             fontfamily='monospace', y=1.01)
plt.tight_layout()
plt.show()

# ─────────────────────────────────────────────
# Ringkasan statistik
# ─────────────────────────────────────────────
print("\n── Statistik Channel ──────────────────────────")
for nama_ch, ori, mod in zip(nama_channel, channels_ori, channels_modif):
    print(f"  {nama_ch:5s} │ mean asli={ori.mean():.1f}  →  mean modif={mod.mean():.1f}"
          f"  │ Δ = {mod.mean()-ori.mean():+.1f}")
````

### **Soal 3 — Deteksi Tepi dan Filter Citra**

* Ambil gambar **objek dengan background bertekstur** (misalnya kain bermotif, jerami, atau batu).
* Terapkan **edge detection (Canny)** dan tampilkan hasilnya.
* Lakukan **thresholding dengan nilai ambang tertentu** (bebas Anda tentukan) agar hanya objek utama yang tersisa.
* Buat **bounding box** di sekitar objek hasil segmentasi (boleh manual atau otomatis).
* Terapkan **filter blur** dan **filter sharpening**, lalu **bandingkan hasil keduanya**.
* Jelaskan bagaimana setiap filter memengaruhi detail gambar dalam format **Markdown**.


````python
# ─────────────────────────────────────────────
# SOAL 3 — Deteksi Tepi dan Filter Citra
# ─────────────────────────────────────────────
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# ── Konfigurasi path ──────────────────────────
IMAGE_PATH = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\canting.jpg"
OUTPUT_DIR = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\output"
os.makedirs(OUTPUT_DIR, exist_ok=True)

nama = "Canting dan Batik"

# ═════════════════════════════════════════════
# HELPER — style axis
# ═════════════════════════════════════════════
def style_ax(ax, judul):
    ax.set_title(judul, color='white', fontsize=10,
                 fontweight='bold', fontfamily='monospace')
    ax.axis('off')

# ═════════════════════════════════════════════
# 1. LOAD GAMBAR
# ═════════════════════════════════════════════
img_bgr = cv2.imread(IMAGE_PATH)
if img_bgr is None:
    raise FileNotFoundError(f"File tidak ditemukan: {IMAGE_PATH}")

img_rgb  = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
print(f"✔ Gambar dimuat: {img_bgr.shape[1]}×{img_bgr.shape[0]}")

# ═════════════════════════════════════════════
# 2. EDGE DETECTION — Canny
# ═════════════════════════════════════════════
# Gaussian blur dulu agar noise tidak terbaca sebagai tepi
blur_for_edge = cv2.GaussianBlur(img_gray, (5, 5), 0)
canny_low     = cv2.Canny(blur_for_edge, threshold1=50,  threshold2=150)
canny_high    = cv2.Canny(blur_for_edge, threshold1=100, threshold2=200)

print("✔ Canny edge detection selesai")

# ═════════════════════════════════════════════
# 3. THRESHOLDING — isolasi objek utama
# ═════════════════════════════════════════════
# Otsu threshold otomatis mencari nilai ambang terbaik
_, thresh_otsu = cv2.threshold(img_gray, 0, 255,
                               cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# Manual threshold (nilai bebas)
THRESH_VAL = 120
_, thresh_manual = cv2.threshold(img_gray, THRESH_VAL, 255, cv2.THRESH_BINARY)

# Adaptive threshold — lebih baik untuk background bertekstur
thresh_adaptive = cv2.adaptiveThreshold(img_gray, 255,
                                        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                        cv2.THRESH_BINARY, 11, 2)

print(f"✔ Thresholding selesai  (manual={THRESH_VAL}, otsu=auto, adaptive=gaussian)")

# ═════════════════════════════════════════════
# 4. BOUNDING BOX — dari kontur hasil threshold
# ═════════════════════════════════════════════
# Morphological closing agar kontur lebih rapat
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (7, 7))
thresh_close = cv2.morphologyEx(thresh_otsu, cv2.MORPH_CLOSE, kernel, iterations=2)

contours, _ = cv2.findContours(thresh_close, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Ambil kontur terbesar (objek utama)
img_bbox = img_rgb.copy()
if contours:
    c_terbesar = max(contours, key=cv2.contourArea)
    x, y, w, h = cv2.boundingRect(c_terbesar)
    cv2.rectangle(img_bbox, (x, y), (x+w, y+h), (0, 255, 100), 3)
    cv2.putText(img_bbox, f"Objek Utama [{w}x{h}]",
                (x, max(y-10, 10)), cv2.FONT_HERSHEY_SIMPLEX,
                0.7, (0, 255, 100), 2)
    print(f"✔ Bounding box objek utama: x={x} y={y} w={w} h={h}")
else:
    print("⚠ Tidak ada kontur ditemukan")

# ═════════════════════════════════════════════
# 5. FILTER — Blur & Sharpening
# ═════════════════════════════════════════════
# ── Blur ──────────────────────────────────────
blur_gaussian = cv2.GaussianBlur(img_rgb, (15, 15), 0)
blur_median   = cv2.medianBlur(img_bgr, 15)
blur_median   = cv2.cvtColor(blur_median, cv2.COLOR_BGR2RGB)

# ── Sharpening ────────────────────────────────
kernel_sharp = np.array([[ 0, -1,  0],
                          [-1,  5, -1],
                          [ 0, -1,  0]], dtype=np.float32)
img_sharp = cv2.filter2D(img_rgb, -1, kernel_sharp)

# Unsharp masking — sharpening lebih halus & natural
blur_unsharp  = cv2.GaussianBlur(img_rgb, (9, 9), 10)
img_unsharp   = cv2.addWeighted(img_rgb, 1.5, blur_unsharp, -0.5, 0)

print("✔ Filter blur dan sharpening selesai")

# ═════════════════════════════════════════════
# 6. SIMPAN SEMUA HASIL
# ═════════════════════════════════════════════
saves = {
    "s3_canny_low.png"      : canny_low,
    "s3_canny_high.png"     : canny_high,
    "s3_thresh_otsu.png"    : thresh_otsu,
    "s3_thresh_manual.png"  : thresh_manual,
    "s3_thresh_adaptive.png": thresh_adaptive,
    "s3_bbox.png"           : cv2.cvtColor(img_bbox, cv2.COLOR_RGB2BGR),
    "s3_blur_gaussian.png"  : cv2.cvtColor(blur_gaussian, cv2.COLOR_RGB2BGR),
    "s3_blur_median.png"    : cv2.cvtColor(blur_median,   cv2.COLOR_RGB2BGR),
    "s3_sharp_kernel.png"   : cv2.cvtColor(img_sharp,     cv2.COLOR_RGB2BGR),
    "s3_sharp_unsharp.png"  : cv2.cvtColor(img_unsharp,   cv2.COLOR_RGB2BGR),
}
for fname, mat in saves.items():
    cv2.imwrite(os.path.join(OUTPUT_DIR, fname), mat)
print(f"✔ Semua file tersimpan di: {OUTPUT_DIR}")

# ═════════════════════════════════════════════
# 7. VISUALISASI
# ═════════════════════════════════════════════

# ── Fig 1: Edge Detection ─────────────────────
fig1, axes = plt.subplots(1, 3, figsize=(18, 6))
fig1.patch.set_facecolor('#1a1a2e')
for ax, img, ttl in zip(axes,
    [img_gray, canny_low, canny_high],
    ["Grayscale", "Canny (50–150)", "Canny (100–200)"]):
    ax.imshow(img, cmap='gray' if img.ndim == 2 else None)
    style_ax(ax, ttl)
plt.suptitle("Soal 3 — Deteksi Tepi Canny",
             color='white', fontsize=14, fontweight='bold',
             fontfamily='monospace')
plt.tight_layout()
plt.show()

# ── Fig 2: Thresholding + BBox ────────────────
fig2, axes = plt.subplots(1, 4, figsize=(22, 6))
fig2.patch.set_facecolor('#1a1a2e')
for ax, img, ttl in zip(axes,
    [img_gray, thresh_manual, thresh_otsu, img_bbox],
    [f"Grayscale", f"Threshold Manual ({THRESH_VAL})",
     "Threshold Otsu (auto)", "Bounding Box Objek"]):
    ax.imshow(img, cmap='gray' if img.ndim == 2 else None)
    style_ax(ax, ttl)
plt.suptitle("Soal 3 — Thresholding & Bounding Box",
             color='white', fontsize=14, fontweight='bold',
             fontfamily='monospace')
plt.tight_layout()
plt.show()

# ── Fig 3: Filter Blur vs Sharp ───────────────
fig3, axes = plt.subplots(1, 5, figsize=(28, 6))
fig3.patch.set_facecolor('#1a1a2e')
for ax, img, ttl in zip(axes,
    [img_rgb, blur_gaussian, blur_median, img_sharp, img_unsharp],
    ["Original", "Blur Gaussian (15×15)",
     "Blur Median (15)", "Sharpen Kernel", "Unsharp Masking"]):
    ax.imshow(img)
    style_ax(ax, ttl)
plt.suptitle("Soal 3 — Perbandingan Filter",
             color='white', fontsize=14, fontweight='bold',
             fontfamily='monospace')
plt.tight_layout()
plt.show()

# ═════════════════════════════════════════════
# 8. PENJELASAN MARKDOWN
# ═════════════════════════════════════════════
md = f"""
# Soal 3 — Deteksi Tepi dan Filter Citra
**Nama:** {nama}

---

## 🔍 Canny Edge Detection
| Parameter | Nilai | Efek |
|-----------|-------|------|
| threshold1 (low) | 50 | Menangkap lebih banyak tepi lemah |
| threshold2 (high) | 150 | Batas tepi kuat |
| threshold1 (low) | 100 | Tepi lemah diabaikan |
| threshold2 (high) | 200 | Hanya tepi sangat kuat yang tersisa |

- Sebelum Canny diterapkan **Gaussian Blur 5×5** untuk meredam noise.
- Threshold rendah (50–150) menghasilkan tepi lebih banyak termasuk tekstur background.
- Threshold tinggi (100–200) menghasilkan tepi bersih hanya pada objek dominan.

---

## 🎚️ Thresholding
| Metode | Nilai Ambang | Keterangan |
|--------|-------------|------------|
| Manual | {THRESH_VAL} | Ditentukan secara eksplisit |
| Otsu | Otomatis | Dihitung dari histogram citra |
| Adaptive | Per-region | Cocok untuk background bertekstur |

- **Otsu** paling efektif untuk gambar dengan distribusi bimodal (objek terang vs gelap).
- **Adaptive** unggul saat pencahayaan tidak merata atau background bertekstur.

---

## 📦 Bounding Box
- Dibuat dari **kontur terbesar** hasil Otsu threshold + morphological closing.
- Morphological closing (kernel 7×7, 2 iterasi) menutup celah kecil pada objek
  sehingga kontur lebih utuh.

---

## 🌫️ Filter Blur
| Filter | Kernel | Efek pada Detail |
|--------|--------|-----------------|
| Gaussian Blur | 15×15 | Menghaluskan secara merata, detail halus hilang |
| Median Blur | 15 | Mempertahankan tepi, efektif hilangkan noise salt-pepper |

- **Gaussian Blur** bekerja dengan rata-rata tertimbang piksel sekitar —
  semakin besar kernel, semakin halus gambar namun detail makin kabur.
- **Median Blur** mengganti nilai piksel dengan median tetangganya —
  lebih tahan terhadap noise impulsif dan tepi objek lebih terjaga.

---

## ✨ Filter Sharpening
| Filter | Metode | Efek pada Detail |
|--------|--------|-----------------|
| Sharpen Kernel | Konvolusi 3×3 | Memperkuat tepi secara langsung |
| Unsharp Masking | Weighted subtract | Sharpening natural, tidak over-enhance |

- **Sharpen Kernel** (nilai tengah +5, tetangga -1) memperkuat perbedaan intensitas
  antar piksel — tekstur dan tepi menjadi lebih tajam namun bisa memunculkan artefak.
- **Unsharp Masking** mengurangi versi blur dari gambar asli, hasilnya lebih natural
  dan dipakai luas di fotografi digital.

---

## 📊 Kesimpulan Pengaruh Filter
| Aspek | Blur Gaussian | Blur Median | Sharpen Kernel | Unsharp Masking |
|-------|:---:|:---:|:---:|:---:|
| Detail halus | ❌ Hilang | ⚠️ Berkurang | ✅ Meningkat | ✅ Meningkat |
| Ketajaman tepi | ❌ Menurun | ⚠️ Terjaga | ✅ Meningkat | ✅ Meningkat |
| Noise | ✅ Berkurang | ✅ Berkurang | ❌ Meningkat | ⚠️ Sedikit naik |
| Artefak | Rendah | Rendah | Tinggi | Rendah |
"""
print(md)
````

### **Soal 4 — Deteksi Wajah dan Filter Digital Kreatif**

* Ambil gambar diri Anda dengan ekspresi wajah **netral**.
* Lakukan **deteksi wajah dan landmark** menggunakan salah satu dari:

  * **MediaPipe**, atau
  * **Dlib**, atau
  * **OpenCV**.
* Buat **overlay filter digital kreatif** karya Anda sendiri, misalnya:

  * topi, kumis, masker, helm, aksesoris, atau bentuk unik lainnya.
  * Filter boleh dibuat dari **gambar eksternal (PNG)** *atau* digambar langsung (misal bentuk lingkaran, garis, poligon, dll).
* Pastikan posisi overlay menyesuaikan **landmark wajah** dengan logis.
* **Gunakan alpha blending sebagai saran** agar hasil tampak lebih natural.
* Tampilkan perbandingan antara **gambar asli** dan **hasil dengan filter**.
* Jelaskan bagaimana Anda menghitung posisi overlay dan tantangan yang dihadapi selama implementasi (gunakan **Markdown**).

````python
# ─────────────────────────────────────────────
# SOAL 4 — Deteksi Wajah dan Filter Digital Kreatif
# Menggunakan MediaPipe Face Mesh
# ─────────────────────────────────────────────
import cv2
import numpy as np
import matplotlib.pyplot as plt
import mediapipe as mp
import os
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ── Install mediapipe jika belum ada ──────────
# pip install mediapipe

# ── Konfigurasi ──────────────────────────────
IMAGE_PATH = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\selfie.jpg"
OUTPUT_DIR = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\output"
os.makedirs(OUTPUT_DIR, exist_ok=True)
nama = "Fajar Rama Adi"

# ═════════════════════════════════════════════
# HELPER FUNCTIONS
# ═════════════════════════════════════════════
def get_landmark_px(landmarks, idx, w, h):
    """Konversi landmark normalized → pixel coordinate."""
    lm = landmarks[idx]
    return int(lm.x * w), int(lm.y * h)

def alpha_blend(background, overlay_bgra, x, y):
    """
    Tempel overlay_bgra (BGRA) ke background (BGR) di posisi (x,y).
    Otomatis crop jika overlay keluar batas gambar.
    """
    bg = background.copy()
    oh, ow = overlay_bgra.shape[:2]
    bh, bw = bg.shape[:2]

    # Hitung area irisan
    x1, y1 = max(x, 0), max(y, 0)
    x2, y2 = min(x + ow, bw), min(y + oh, bh)
    if x2 <= x1 or y2 <= y1:
        return bg  # tidak ada irisan

    # Crop overlay sesuai irisan
    ox1, oy1 = x1 - x, y1 - y
    ox2, oy2 = ox1 + (x2 - x1), oy1 + (y2 - y1)
    roi     = bg[y1:y2, x1:x2]
    ov_crop = overlay_bgra[oy1:oy2, ox1:ox2]

    alpha = ov_crop[:, :, 3:4].astype(np.float32) / 255.0
    ov_bgr = ov_crop[:, :, :3].astype(np.float32)
    roi_f  = roi.astype(np.float32)

    blended = (alpha * ov_bgr + (1 - alpha) * roi_f).astype(np.uint8)
    bg[y1:y2, x1:x2] = blended
    return bg

def style_ax(ax, judul):
    ax.set_title(judul, color='white', fontsize=11,
                 fontweight='bold', fontfamily='monospace')
    ax.axis('off')

# ═════════════════════════════════════════════
# 1. LOAD GAMBAR
# ═════════════════════════════════════════════
img_bgr = cv2.imread(IMAGE_PATH)
if img_bgr is None:
    raise FileNotFoundError(f"File tidak ditemukan: {IMAGE_PATH}")

img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
H, W    = img_bgr.shape[:2]
print(f"✔ Gambar dimuat: {W}×{H}")

# ═════════════════════════════════════════════
# 2. DETEKSI WAJAH — MediaPipe Face Mesh
# ═════════════════════════════════════════════
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# Download model dulu (jalankan sekali)
import urllib.request
MODEL_PATH = "face_landmarker.task"
urllib.request.urlretrieve(
    "https://storage.googleapis.com/mediapipe-models/face_landmarker/face_landmarker/float16/1/face_landmarker.task",
    MODEL_PATH
)

base_options = mp.tasks.BaseOptions(model_asset_path=MODEL_PATH)
options = vision.FaceLandmarkerOptions(
    base_options=base_options,
    output_face_blendshapes=False,
    num_faces=1
)

with vision.FaceLandmarker.create_from_options(options) as detector:
    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=img_rgb
    )
    results = detector.detect(mp_image)

if not results.face_landmarks:
    raise RuntimeError("Wajah tidak terdeteksi!")

landmarks = results.face_landmarks[0]
print(f"✔ Wajah terdeteksi — {len(landmarks)} landmark")

# ─── Landmark index penting (MediaPipe 468 points) ───
# Referensi: https://mediapipe.dev/images/mobile/face_mesh_full_rerun.gif
IDX = {
    "hidung_tip"    : 1,
    "hidung_bridge" : 6,
    "dahi"          : 10,
    "dagu"          : 152,
    "mata_kiri_luar": 33,
    "mata_kiri_dalam": 133,
    "mata_kanan_luar": 263,
    "mata_kanan_dalam": 362,
    "bibir_atas"    : 13,
    "bibir_bawah"   : 14,
    "pipi_kiri"     : 234,
    "pipi_kanan"    : 454,
    "alis_kiri_L"   : 70,
    "alis_kiri_R"   : 107,
    "alis_kanan_L"  : 336,
    "alis_kanan_R"  : 300,
}

# Konversi semua ke pixel
pts = {k: get_landmark_px(landmarks, v, W, H) for k, v in IDX.items()}

# ─── Ukuran referensi wajah ───
lebar_wajah  = abs(pts["pipi_kanan"][0] - pts["pipi_kiri"][0])
tinggi_wajah = abs(pts["dagu"][1]       - pts["dahi"][1])
print(f"  Lebar wajah  : {lebar_wajah}px")
print(f"  Tinggi wajah : {tinggi_wajah}px")

# ═════════════════════════════════════════════
# 3. GAMBAR LANDMARK DEBUG (opsional — tampilkan titik)
# ═════════════════════════════════════════════
img_landmark = img_rgb.copy()
for name, (px, py) in pts.items():
    cv2.circle(img_landmark, (px, py), 5, (0, 255, 100), -1)
    cv2.putText(img_landmark, name, (px+6, py),
                cv2.FONT_HERSHEY_SIMPLEX, 0.3, (255, 255, 0), 1)

# ═════════════════════════════════════════════
# 4. BUAT OVERLAY FILTER — digambar dengan NumPy/OpenCV
#    Filter: Topi koboi + Kacamata bulat + Kumis + Pipi merah
# ═════════════════════════════════════════════

def buat_topi_koboi(lebar, tinggi):
    """
    Topi koboi sederhana: brim (lebar) + crown (tinggi).
    Return BGRA canvas.
    """
    cw = int(lebar * 1.8)
    ch = int(tinggi * 0.7)
    canvas = np.zeros((ch, cw, 4), dtype=np.uint8)

    warna_topi  = (30,  80, 139, 255)   # coklat tua (BGR)
    warna_band  = (20,  50,  90, 255)   # band lebih gelap
    warna_tepi  = (10,  30,  60, 255)

    # ── Brim (pinggir topi) ───────────────────
    brim_h = int(ch * 0.28)
    brim_y = ch - brim_h
    pts_brim = np.array([
        [0,          ch],
        [int(cw*0.08), brim_y],
        [int(cw*0.92), brim_y],
        [cw,         ch],
    ], np.int32)
    cv2.fillPoly(canvas, [pts_brim], warna_topi)
    cv2.polylines(canvas, [pts_brim], True, warna_tepi, 2)

    # ── Crown (badan topi) ────────────────────
    crown_x1 = int(cw * 0.28)
    crown_x2 = int(cw * 0.72)
    crown_y1 = 0
    crown_y2 = brim_y + 4
    pts_crown = np.array([
        [crown_x1 + int((crown_x2-crown_x1)*0.08), crown_y1],
        [crown_x2 - int((crown_x2-crown_x1)*0.08), crown_y1],
        [crown_x2, crown_y2],
        [crown_x1, crown_y2],
    ], np.int32)
    cv2.fillPoly(canvas, [pts_crown], warna_topi)

    # ── Band topi ─────────────────────────────
    band_y1 = brim_y - int(ch * 0.12)
    band_y2 = brim_y + 2
    cv2.rectangle(canvas,
                  (crown_x1, band_y1), (crown_x2, band_y2),
                  warna_band, -1)

    # ── Highlight ─────────────────────────────
    cv2.line(canvas,
             (crown_x1 + 8, int(ch*0.1)),
             (crown_x2 - 8, int(ch*0.1)),
             (60, 120, 180, 180), 3)
    return canvas


def buat_kacamata(jarak_mata, r):
    """
    Kacamata bulat dua lensa.
    jarak_mata = jarak antar pusat mata, r = radius lensa
    Return BGRA canvas, anchor = tengah atas canvas
    """
    pad  = r + 10
    cw   = jarak_mata + 2 * pad
    ch   = 2 * r + 2 * pad
    canvas = np.zeros((ch, cw, 4), dtype=np.uint8)

    warna_frame = (20, 20, 20, 240)
    warna_lensa = (200, 230, 180, 60)   # hijau transparan

    cy = ch // 2
    cx_kiri  = pad
    cx_kanan = pad + jarak_mata

    for cx in [cx_kiri, cx_kanan]:
        # Lensa
        cv2.circle(canvas, (cx, cy), r, warna_lensa, -1)
        # Frame
        cv2.circle(canvas, (cx, cy), r, warna_frame, 3)

    # Batang tengah
    cv2.line(canvas, (cx_kiri+r, cy), (cx_kanan-r, cy), warna_frame, 3)
    # Batang samping kiri
    cv2.line(canvas, (cx_kiri-r, cy), (0, cy-5), warna_frame, 3)
    # Batang samping kanan
    cv2.line(canvas, (cx_kanan+r, cy), (cw-1, cy-5), warna_frame, 3)

    return canvas, cx_kiri, cx_kanan, cy


def buat_kumis(lebar):
    """
    Kumis melengkung dari polyline.
    Return BGRA canvas.
    """
    cw = lebar
    ch = int(lebar * 0.35)
    canvas = np.zeros((ch, cw, 4), dtype=np.uint8)

    warna = (20, 20, 20, 230)
    cx = cw // 2

    # Sisi kiri
    pts_kiri = np.array([
        [cx,        int(ch*0.2)],
        [int(cx*0.6), int(ch*0.1)],
        [int(cx*0.1), int(ch*0.5)],
        [int(cx*0.3), int(ch*0.9)],
        [cx,         int(ch*0.6)],
    ], np.int32)
    cv2.fillPoly(canvas, [pts_kiri], warna)

    # Sisi kanan (mirror)
    pts_kanan = pts_kiri.copy()
    pts_kanan[:, 0] = cw - pts_kiri[:, 0]
    cv2.fillPoly(canvas, [pts_kanan], warna)

    return canvas


def buat_pipi_merah(r):
    """Lingkaran pipi merah transparan."""
    cw = ch = r * 2 + 4
    canvas = np.zeros((ch, cw, 4), dtype=np.uint8)
    cv2.circle(canvas, (r+2, r+2), r, (80, 80, 220, 100), -1)   # merah muda
    return canvas

# ═════════════════════════════════════════════
# 5. POSISIKAN OVERLAY KE GAMBAR
# ═════════════════════════════════════════════
img_filter = cv2.cvtColor(img_rgb.copy(), cv2.COLOR_RGB2BGR)

# ── A. TOPI KOBOI ─────────────────────────────
topi = buat_topi_koboi(lebar_wajah, tinggi_wajah)
th, tw = topi.shape[:2]
# Anchor: tengah bawah topi = titik dahi
topi_x = pts["dahi"][0] - tw // 2
topi_y = pts["dahi"][1] - int(th * 0.88)
img_filter = alpha_blend(img_filter, topi, topi_x, topi_y)
print("✔ Overlay topi selesai")

# ── B. KACAMATA ───────────────────────────────
jarak_mata = abs(pts["mata_kanan_luar"][0] - pts["mata_kiri_luar"][0])
r_lensa    = int(jarak_mata * 0.22)
kacamata, cx_kiri, cx_kanan, cy_lensa = buat_kacamata(jarak_mata, r_lensa)
kh, kw = kacamata.shape[:2]

# Pusat kacamata = rata-rata Y kedua mata
mata_y   = (pts["mata_kiri_luar"][1] + pts["mata_kanan_luar"][1]) // 2
kacamata_x = pts["mata_kiri_luar"][0] - cx_kiri - r_lensa
kacamata_y = mata_y - cy_lensa
img_filter = alpha_blend(img_filter, kacamata, kacamata_x, kacamata_y)
print("✔ Overlay kacamata selesai")

# ── C. KUMIS ──────────────────────────────────
kumis_lebar = int(lebar_wajah * 0.55)
kumis = buat_kumis(kumis_lebar)
mh, mw = kumis.shape[:2]
# Anchor: tengah kumis = tepat di bawah hidung
kumis_x = pts["hidung_tip"][0] - mw // 2
kumis_y = pts["hidung_tip"][1] + 4
img_filter = alpha_blend(img_filter, kumis, kumis_x, kumis_y)
print("✔ Overlay kumis selesai")

# ── D. PIPI MERAH ─────────────────────────────
r_pipi = int(lebar_wajah * 0.12)
pipi   = buat_pipi_merah(r_pipi)
# Pipi kiri
img_filter = alpha_blend(img_filter, pipi,
                          pts["pipi_kiri"][0]  - r_pipi,
                          pts["pipi_kiri"][1]  - r_pipi)
# Pipi kanan
img_filter = alpha_blend(img_filter, pipi,
                          pts["pipi_kanan"][0] - r_pipi,
                          pts["pipi_kanan"][1] - r_pipi)
print("✔ Overlay pipi selesai")

# ═════════════════════════════════════════════
# 6. SIMPAN HASIL
# ═════════════════════════════════════════════
cv2.imwrite(os.path.join(OUTPUT_DIR, "s4_landmark.png"),
            cv2.cvtColor(img_landmark, cv2.COLOR_RGB2BGR))
cv2.imwrite(os.path.join(OUTPUT_DIR, "s4_filter.png"), img_filter)
print(f"✔ File tersimpan di: {OUTPUT_DIR}")

# ═════════════════════════════════════════════
# 7. VISUALISASI
# ═════════════════════════════════════════════
fig, axes = plt.subplots(1, 3, figsize=(21, 8))
fig.patch.set_facecolor('#1a1a2e')

for ax, img, ttl in zip(axes,
    [img_rgb,
     img_landmark,
     cv2.cvtColor(img_filter, cv2.COLOR_BGR2RGB)],
    ["Gambar Asli",
     "Landmark MediaPipe",
     "Filter Kreatif (Topi+Kacamata+Kumis+Pipi)"]):
    ax.imshow(img)
    style_ax(ax, ttl)

plt.suptitle(f"Soal 4 — Deteksi Wajah & Filter Digital  |  {nama}",
             color='white', fontsize=14, fontweight='bold',
             fontfamily='monospace')
plt.tight_layout()
plt.show()

# ═════════════════════════════════════════════
# 8. PENJELASAN MARKDOWN
# ═════════════════════════════════════════════
md = f"""
# Soal 4 — Deteksi Wajah dan Filter Digital Kreatif
**Nama:** {nama}

---

## 🧠 Metode Deteksi — MediaPipe Face Mesh

MediaPipe Face Mesh mendeteksi **468 landmark** wajah dalam koordinat
ternormalisasi (0.0–1.0). Konversi ke piksel dilakukan dengan:

```python
lm = landmarks[idx]
px = int(lm.x * W)
py = int(lm.y * H)
```

## 🎨 Filter yang Diterapkan

| Filter     | Anchor Landmark         | Teknik                  |
|------------|-------------------------|-------------------------|
| Topi Koboi | `dahi` (idx 10)         | `fillPoly` + alpha blend |
| Kacamata   | `mata_kiri/kanan_luar`  | `circle` + alpha blend  |
| Kumis      | `hidung_tip` (idx 1)    | `fillPoly` mirror       |
| Pipi Merah | `pipi_kiri/kanan`       | `circle` transparan     |

## 📁 Output
- `s4_landmark.png` — visualisasi titik landmark
- `s4_filter.png`   — hasil filter kreatif
"""  # ← INI YANG HILANG

print(md)
````

````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# ── Konfigurasi ──────────────────────────────
IMAGE_PATH = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\karya.jpg"
OUTPUT_DIR = r"C:\Users\ACER\Documents\citra\Pengolahan_Multimedia\data\output"
os.makedirs(OUTPUT_DIR, exist_ok=True)
nama = "Fajar Rama Adi"

# ═════════════════════════════════════════════
# HELPER — tampilan subplot
# ═════════════════════════════════════════════
def style_ax(ax, judul, cmap=None):
    ax.set_title(judul, color='white', fontsize=10,
                 fontweight='bold', fontfamily='monospace', pad=8)
    ax.axis('off')

# ═════════════════════════════════════════════
# 1. LOAD GAMBAR ASLI
# ═════════════════════════════════════════════
img_bgr = cv2.imread(IMAGE_PATH)
if img_bgr is None:
    raise FileNotFoundError(f"File tidak ditemukan: {IMAGE_PATH}")

img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
H, W    = img_bgr.shape[:2]
print(f"✔ Gambar dimuat: {W}×{H}")

# ═════════════════════════════════════════════
# 2. KONVERSI KE GRAYSCALE
# ═════════════════════════════════════════════
gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
print("✔ Konversi grayscale selesai")

# ═════════════════════════════════════════════
# 3. KOREKSI PERSPEKTIF — Tentukan 4 Titik Manual
# ─────────────────────────────────────────────
# ⚠ SESUAIKAN koordinat src_pts dengan sudut dokumen/objek di foto Anda!
#
# Urutan titik (konsisten, searah jarum jam dari kiri-atas):
#   [0] kiri-atas   (TL)
#   [1] kanan-atas  (TR)
#   [2] kanan-bawah (BR)
#   [3] kiri-bawah  (BL)
#
# Cara mendapatkan koordinat:
#   Jalankan skrip bantuan di bawah (bagian 3b) untuk klik interaktif,
#   lalu salin nilai yang muncul di terminal ke src_pts.
# ─────────────────────────────────────────────

# ── 3a. Koordinat sumber (sudut objek di foto asli) ──
#    Ganti angka ini sesuai foto Anda!
src_pts = np.float32([
    [120,  85],   # kiri-atas
    [780,  60],   # kanan-atas
    [810, 570],   # kanan-bawah
    [ 90, 590],   # kiri-bawah
])

# ── 3b. Skrip klik interaktif (jalankan terpisah jika perlu) ──────────
# import cv2, numpy as np
# pts = []
# def klik(event, x, y, flags, param):
#     if event == cv2.EVENT_LBUTTONDOWN:
#         pts.append([x, y])
#         print(f"  Titik {len(pts)}: [{x}, {y}]")
#         cv2.circle(param, (x,y), 6, (0,255,0), -1)
#         cv2.imshow("Klik 4 sudut", param)
#         if len(pts) == 4:
#             cv2.destroyAllWindows()
# preview = cv2.imread(IMAGE_PATH)
# cv2.namedWindow("Klik 4 sudut")
# cv2.setMouseCallback("Klik 4 sudut", klik, preview)
# cv2.imshow("Klik 4 sudut", preview)
# cv2.waitKey(0)
# print("src_pts =", pts)
# ────────────────────────────────────────────────────────────────────

# ── Ukuran output setelah koreksi ────────────
lebar_out  = 600
tinggi_out = 800

dst_pts = np.float32([
    [0,          0],
    [lebar_out,  0],
    [lebar_out,  tinggi_out],
    [0,          tinggi_out],
])

# ── Hitung matriks homografi & warp ──────────
M, _ = cv2.findHomography(src_pts, dst_pts)
warped = cv2.warpPerspective(img_bgr, M, (lebar_out, tinggi_out))
warped_gray = cv2.warpPerspective(gray, M, (lebar_out, tinggi_out))
warped_rgb  = cv2.cvtColor(warped, cv2.COLOR_BGR2RGB)
print("✔ Koreksi perspektif selesai")

# ── Gambar titik src di gambar asli ──────────
img_with_pts = img_rgb.copy()
for i, (px, py) in enumerate(src_pts.astype(int)):
    cv2.circle(img_with_pts, (px, py), 10, (255, 80, 0), -1)
    cv2.putText(img_with_pts, f"P{i+1}", (px+12, py),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 220, 0), 2)
label = ["TL", "TR", "BR", "BL"]
for i in range(4):
    cv2.line(img_with_pts,
             tuple(src_pts[i].astype(int)),
             tuple(src_pts[(i+1)%4].astype(int)),
             (0, 255, 100), 2)

# ═════════════════════════════════════════════
# 4. THRESHOLDING — Pilihan: ADAPTIF
# ─────────────────────────────────────────────
# Alasan memilih Adaptive Thresholding (bukan Otsu):
#   • Gambar hasil scan/foto sering memiliki pencahayaan TIDAK MERATA
#     (bayangan, vignette, pantulan cahaya). Otsu menghitung satu nilai
#     threshold global → area gelap menjadi hitam semua / area terang
#     menjadi putih semua.
#   • Adaptive Thresholding menghitung threshold LOKAL per blok piksel,
#     sehingga teks/garis tetap terbaca meski pencahayaan tidak rata.
#   • Untuk dokumen hasil foto dengan bayangan, adaptive jauh lebih andal.
# ─────────────────────────────────────────────

# ── Gaussian Adaptive Thresholding ───────────
thresh_adaptive = cv2.adaptiveThreshold(
    warped_gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,   # bobot Gaussian per blok
    cv2.THRESH_BINARY,
    blockSize=31,    # ukuran blok lokal (harus ganjil)
    C=10             # konstanta dikurangi dari mean lokal
)
print("✔ Adaptive thresholding selesai")

# ── Otsu (untuk perbandingan — opsional) ─────
_, thresh_otsu = cv2.threshold(
    warped_gray, 0, 255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)
otsu_val = cv2.threshold(warped_gray, 0, 255,
                          cv2.THRESH_BINARY + cv2.THRESH_OTSU)[0]
print(f"  Nilai threshold Otsu (referensi): {otsu_val:.1f}")

# ═════════════════════════════════════════════
# 5. SIMPAN SEMUA HASIL
# ═════════════════════════════════════════════
cv2.imwrite(os.path.join(OUTPUT_DIR, "s5_gray.png"),    gray)
cv2.imwrite(os.path.join(OUTPUT_DIR, "s5_warped.png"),  warped)
cv2.imwrite(os.path.join(OUTPUT_DIR, "s5_warped_gray.png"), warped_gray)
cv2.imwrite(os.path.join(OUTPUT_DIR, "s5_adaptive.png"), thresh_adaptive)
cv2.imwrite(os.path.join(OUTPUT_DIR, "s5_otsu.png"),    thresh_otsu)
print(f"✔ Semua file tersimpan di: {OUTPUT_DIR}")

# ═════════════════════════════════════════════
# 6. VISUALISASI — 5 PANEL GRID
# ═════════════════════════════════════════════
fig, axes = plt.subplots(2, 3, figsize=(22, 14))
fig.patch.set_facecolor('#0d1117')
axes = axes.flatten()

panels = [
    (img_with_pts,    "1. Asli + 4 Titik Perspektif",    None),
    (gray,            "2. Grayscale",                     'gray'),
    (warped_rgb,      "3. Hasil Koreksi Perspektif",      None),
    (warped_gray,     "4. Warped Grayscale",              'gray'),
    (thresh_adaptive, "5. Adaptive Threshold (PILIHAN)",  'gray'),
    (thresh_otsu,     f"6. Otsu Threshold (ref T={otsu_val:.0f})", 'gray'),
]

for ax, (img, ttl, cmap) in zip(axes, panels):
    if cmap:
        ax.imshow(img, cmap=cmap, vmin=0, vmax=255)
    else:
        ax.imshow(img)
    style_ax(ax, ttl)

plt.suptitle(
    f"Soal 5 — Perspektif & Peningkatan Kualitas Citra  |  {nama}",
    color='white', fontsize=14, fontweight='bold',
    fontfamily='monospace', y=1.01
)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "s5_grid.png"),
            dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
plt.show()
print("✔ Grid visualisasi tersimpan: s5_grid.png")

# ═════════════════════════════════════════════
# 7. PENJELASAN MARKDOWN
# ═════════════════════════════════════════════
md = f"""
# Soal 5 — Perspektif dan Peningkatan Kualitas Citra
**Nama:** {nama}

---

## 📐 Tahap Pemrosesan

### Tahap 1 — Konversi Grayscale
`cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)`

Mengubah citra 3-channel (BGR) menjadi 1-channel (0–255).
- Mengurangi dimensi data → pemrosesan lebih cepat.
- Threshold dan deteksi tepi bekerja pada intensitas tunggal, bukan warna.
- Formula: `Y = 0.114·B + 0.587·G + 0.299·R` (bobot persepsi manusia).

---

### Tahap 2 — Koreksi Perspektif (Homografi)
`cv2.findHomography(src_pts, dst_pts)` → `cv2.warpPerspective(...)`

Transformasi proyektif (homografi) memetakan 4 titik sudut objek yang terdistorsi
ke persegi panjang ideal. Matriks 3×3 dihitung dari pasangan titik sumber–tujuan.

```
⎡ h00 h01 h02 ⎤   ⎡ x ⎤   ⎡ x' ⎤
⎢ h10 h11 h12 ⎥ × ⎢ y ⎥ → ⎢ y' ⎥  (koordinat homogen)
⎣ h20 h21 h22 ⎦   ⎣ 1 ⎦   ⎣ w  ⎦
```

Manfaat: menghilangkan distorsi sudut pandang sehingga teks/gambar terlihat
sejajar seperti dipotret dari depan.

---

### Tahap 3 — Adaptive Thresholding *(dipilih)*
`cv2.adaptiveThreshold(..., ADAPTIVE_THRESH_GAUSSIAN_C, THRESH_BINARY, 31, 10)`

Mengapa **Adaptive**, bukan Otsu?

| Kriteria           | Adaptive Threshold        | Otsu Threshold         |
|--------------------|---------------------------|------------------------|
| Jenis threshold    | Lokal per blok piksel     | Global satu nilai      |
| Pencahayaan merata | ✔ Ya / Tidak              | ✔ Ya                   |
| Pencahayaan tidak merata | ✔ **Sangat baik** | ✘ Kurang andal         |
| Bayangan / vignette | ✔ Ditangani              | ✘ Teks hilang di area gelap |
| Kecepatan          | Lebih lambat              | Lebih cepat            |

Foto objek datar (kertas, papan tulis) hampir selalu memiliki pencahayaan
**tidak merata** karena sudut lampu, bayangan, atau vignette kamera.
Adaptive threshold menghitung rata-rata intensitas dalam blok lokal (31×31 px)
dan mengurangi konstanta C=10, sehingga teks tetap tajam meskipun latar bervariasi.

---

## 🎯 Kesimpulan

Kombinasi tiga tahap ini meningkatkan kualitas visual citra secara signifikan:
1. **Grayscale** → menyederhanakan data.
2. **Homografi** → meluruskan perspektif agar dokumen terbaca proporsional.
3. **Adaptive Threshold** → mempertajam kontras teks/garis pada pencahayaan tidak merata.

Output akhir berupa citra biner hitam-putih yang bersih, cocok untuk OCR,
pengarsipan digital, atau analisis lebih lanjut.
"""

print(md)
````
