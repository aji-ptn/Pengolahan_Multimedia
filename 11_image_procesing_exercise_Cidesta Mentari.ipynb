### **Soal 3 — Deteksi Tepi dan Filter Citra**

* Ambil gambar **objek dengan background bertekstur** (misalnya kain bermotif, jerami, atau batu).
* Terapkan **edge detection (Canny)** dan tampilkan hasilnya.
* Lakukan **thresholding dengan nilai ambang tertentu** (bebas Anda tentukan) agar hanya objek utama yang tersisa.
* Buat **bounding box** di sekitar objek hasil segmentasi (boleh manual atau otomatis).
* Terapkan **filter blur** dan **filter sharpening**, lalu **bandingkan hasil keduanya**.
* Jelaskan bagaimana setiap filter memengaruhi detail gambar dalam format **Markdown**.


```python
img = cv2.imread('assets_ws4/tekstur.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis('off')
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.title("Original")
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Canny Edge")
plt.imshow(edges, cmap='gray')
plt.axis('off')
```

>>Canny berfungsi untuk mendeteksi perubahan intensitas pada gambar sehingga menghasilkan garis tepi objek. Hasilnya menunjukkan struktur utama dari objek dan menghilangkan informasi warna.

```python
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Threshold")
plt.axis('off')
```

```python
_, thresh = cv2.threshold(gray, 130, 260, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Threshold")
plt.axis('off')
```

```python
_, thresh = cv2.threshold(gray, 140, 300, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Threshold")
plt.axis('off')
```

```python
_, thresh = cv2.threshold(gray, 140, 250, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Threshold")
plt.axis('off')
```

```python
_, thresh = cv2.threshold(gray, 135, 250, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Threshold")
plt.axis('off')
```

>>Thresholding digunakan untuk mengubah citra grayscale menjadi citra biner berdasarkan nilai ambang tertentu. Pixel di atas threshold menjadi putih, sedangkan di bawahnya menjadi hitam, sehingga objek utama lebih mudah dipisahkan dari background.

```python
thresh = cv2.adaptiveThreshold(
    gray, 255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11, 2
)

plt.imshow(thresh, cmap='gray')
plt.title("Adaptive Threshold")
plt.axis('off')
```

>>Pada proses thresholding, saya menggunakan dua metode yaitu threshold biasa (global threshold) dan adaptive threshold. Alasan saya menggunakan dua metode ini untuk bahan perbandingan saya dan juga saya tertarik dengan saran AI yang saya gunakan untuk menggunakan proses adaptive threshold agar dapat melihat hasil yang lebih baik antara threshold biasa dengan adaptive threshold.

>>Threshold biasa menggunakan satu nilai ambang yang sama untuk seluruh gambar. Piksel dengan nilai di atas ambang akan menjadi putih, sedangkan yang di bawahnya menjadi hitam. Metode ini cukup sederhana, tetapi kurang efektif jika gambar memiliki pencahayaan yang tidak merata atau background yang kompleks.

>>Sedangkan pada adaptive threshold menentukan nilai ambang berdasarkan kondisi di setiap bagian gambar. Dengan cara ini, metode adaptive threshold lebih mampu menangani variasi pencahayaan dan detail pada gambar yang memiliki banyak tekstur.

>>Dari hasil yang diperoleh, terlihat bahwa threshold biasa kurang mampu memisahkan objek dengan jelas karena banyak bagian yang masih bercampur dengan background. Sementara itu, adaptive threshold memberikan hasil yang lebih baik karena objek utama terlihat lebih jelas dan terpisah dari background.


```python
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

img_box = img_rgb.copy()

for cnt in contours:
    x, y, w, h = cv2.boundingRect(cnt)
    
    # filter biar gak ambil noise kecil
    if w > 50 and h > 50:
        cv2.rectangle(img_box, (x, y), (x+w, y+h), (255, 0, 0), 2)

plt.imshow(img_box)
plt.title("Bounding Box")
plt.axis('off')
```

>>Pada bounding box digunakan untuk mengelilingi objek hasil segmentasi menggunakan koordinat minimum dan maksimum dari kontur. Teknik ini membantu dalam identifikasi lokasi objek secara spasial.

```python
blur = cv2.GaussianBlur(img_rgb, (5,5), 0)

plt.imshow(blur)
plt.title("Blur")
plt.axis('off')
```

```python
blur = cv2.GaussianBlur(img_rgb, (15,15), 0)

plt.imshow(blur)
plt.title("Blur")
plt.axis('off')
```

>>Blur (Gaussian Blur) digunakan untuk menghaluskan citra dengan mengurangi noise dan detail kecil. Namun, efek sampingnya adalah hilangnya ketajaman objek.

```python
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

sharpen = cv2.filter2D(img_rgb, -1, kernel)

plt.imshow(sharpen)
plt.title("Sharpen")
plt.axis('off')
```

>>Sharpening digunakan untuk meningkatkan ketajaman citra dengan memperkuat perbedaan intensitas antar pixel. Hal ini membuat tepi objek menjadi lebih jelas dibandingkan citra asli.

```python
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(1,3,2)
plt.title("Blur")
plt.imshow(blur)
plt.axis('off')

plt.subplot(1,3,3)
plt.title("Sharpen")
plt.imshow(sharpen)
plt.axis('off')
```

>>Pada hasil percobaan, untuk filter blur membuat gambar jadi lebih halus karena detail dan noise berkurang, sehingga objek terlihat agak tidak tajam. Sedangkan sharpen justru kebalikannya, yaitu memperjelas tepi objek dan meningkatkan detail sehingga gambar terlihat lebih tajam dibandingkan original.

>>Jadi, blur dipakai untuk menghaluskan gambar, sedangkan sharpen dipakai untuk menegaskan detail atau edge pada gambar.

```python
import cv2, mediapipe as mp, numpy as np, matplotlib.pyplot as plt

img = cv2.imread("assets_ws4/selfie.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
h, w, _ = img.shape
output = img.copy()
```

```python
mp_face = mp.solutions.face_mesh
face_mesh = mp_face.FaceMesh(static_image_mode=True)

results = face_mesh.process(img)
```

```python
if results.multi_face_landmarks:
    print("Wajah terdeteksi")
else:
    print("Tidak terdeteksi wajah")
```

```python
if results.multi_face_landmarks:
    for face in results.multi_face_landmarks:
        x = int(face.landmark[10].x * w)
        y = int(face.landmark[10].y * h)

        # TOPI
        hat = np.array([
            [x-80, y-20],
            [x+80, y-20],
            [x, y-120]
        ], np.int32)

        cv2.fillPoly(output, [hat], (0, 0, 255))
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(output)
plt.title("Filter Wajah")
plt.axis('off')

plt.show()
```

>>Pada tahap pertama ini, dilakukan pembuatan filter sederhana berupa bentuk segitiga yang ditempatkan pada area dahi. Segitiga ini digunakan sebagai representasi awal dari overlay digital berbasis koordinat landmark wajah. Hasil ini menunjukkan bahwa sistem sudah mampu menempelkan objek sederhana pada posisi wajah.

>>Koordinat landmark dari MediaPipe masih dalam bentuk normalisasi (0–1), sehingga perlu dikonversi ke koordinat pixel dengan rumus:

>>**x pixel = x landmark × lebar_gambar**

>>**y pixel = y landmark × tinggi_gambar**

>>Setelah dikonversi, titik tersebut digunakan untuk menggambar bentuk segitiga pada wajah atau dahi.

```python
if results.multi_face_landmarks:
    for face in results.multi_face_landmarks:

        # titik hidung (nose tip)
        nose = face.landmark[1]
        x = int(nose.x * w)
        y = int(nose.y * h)

        # posisi kumis (sedikit di bawah hidung)
        y_mustache = y + 30

        # bentuk kumis kiri
        cv2.line(output, (x-60, y_mustache), (x-10, y_mustache+10), (0,0,0), 5)

        # tengah kumis
        cv2.line(output, (x-10, y_mustache+10), (x+10, y_mustache+10), (0,0,0), 5)

        # kanan kumis
        cv2.line(output, (x+10, y_mustache+10), (x+60, y_mustache), (0,0,0), 5)
```

Karena ukuran gambar cukup besar, koordinat titik juga harus disesuaikan dengan resolusi gambar agar posisi titik berada tepat pada sudut objek.

```python
points = [
    (520, 420),    # kiri atas
    (3420, 560),   # kanan atas
    (3510, 1820),  # kanan bawah
    (470, 1910)    # kiri bawah
]
```

```python
pts_src = np.float32(points)

width = 3000
height = 1500

pts_dst = np.float32([
    [0, 0],
    [width, 0],
    [width, height],
    [0, height]
])

matrix = cv2.getPerspectiveTransform(pts_src, pts_dst)

warped = cv2.warpPerspective(img, matrix, (width, height))

warped_rgb = cv2.cvtColor(warped, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12,7))
plt.imshow(warped_rgb)
plt.title("Koreksi Perspektif")
plt.axis("off")
plt.show()
```

>>Pada tahap ini dilakukan transformasi homografi untuk memperbaiki distorsi perspektif pada gambar papan tulis. Empat titik sudut yang telah dipilih sebelumnya digunakan sebagai acuan untuk meluruskan objek. Fungsi `getPerspectiveTransform()` digunakan untuk membuat matriks transformasi, sedangkan `warpPerspective()` digunakan untuk menerapkan transformasi tersebut ke citra. Hasilnya papan tulis terlihat lebih lurus, sejajar, dan proporsional sehingga tulisan pada papan menjadi lebih mudah dibaca.

```python
warped_gray = cv2.cvtColor(warped, cv2.COLOR_BGR2GRAY)

# blur ringan
blur = cv2.GaussianBlur(warped_gray, (3,3), 0)

plt.figure(figsize=(10,5))
plt.imshow(blur, cmap='gray')
plt.title("Grayscale")
plt.axis("off")
plt.show()
```

>>Pada tahap grayscale, gambar diubah menjadi citra abu-abu dengan hanya mempertahankan intensitas cahaya. Untuk meningkatkan kualitas visual, digunakan Gaussian Blur ringan agar noise dan detail yang terlalu kasar dapat dikurangi. Hasil grayscale menjadi lebih halus dan nyaman dilihat tanpa menghilangkan informasi penting pada citra. Teknik ini membantu proses preprocessing menjadi lebih stabil sebelum dilakukan thresholding.

```python
adaptive = cv2.adaptiveThreshold(
    blur,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

plt.figure(figsize=(10,5))
plt.imshow(adaptive, cmap='gray')
plt.title("Adaptive Threshold")
plt.axis("off")
plt.show()
```

>>Pada tahap ini digunakan adaptive thresholding untuk mengubah citra grayscale menjadi citra biner. Metode adaptive threshold dipilih karena kondisi pencahayaan pada papan tulis tidak merata akibat pantulan cahaya lampu. Teknik ini menentukan nilai threshold secara lokal pada setiap area gambar sehingga hasil pemisahan tulisan dan latar belakang menjadi lebih baik dibandingkan metode Otsu. Dengan teknik ini tulisan pada papan tulis terlihat lebih jelas dan detail citra menjadi lebih mudah diamati.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# membaca gambar
img = cv2.imread("assets_ws4/object.jpg")

# konversi BGR ke RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# =========================
# 1. GRAYSCALE
# =========================
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# blur ringan agar lebih halus
blur = cv2.GaussianBlur(gray, (3,3), 0)

# =========================
# 2. MENENTUKAN 4 TITIK
# =========================
points = [
    (520, 420),    # kiri atas
    (3420, 560),   # kanan atas
    (3510, 1820),  # kanan bawah
    (470, 1910)    # kiri bawah
]

# menampilkan titik perspektif
img_copy = img_rgb.copy()

for point in points:
    cv2.circle(img_copy, point, 40, (255, 0, 0), -1)

plt.figure(figsize=(12,7))
plt.imshow(img_copy)
plt.title("4 Titik Perspektif")
plt.axis("off")
plt.show()

# =========================
# 3. KOREKSI PERSPEKTIF
# =========================
pts_src = np.float32(points)

width = 3000
height = 1500

pts_dst = np.float32([
    [0, 0],
    [width, 0],
    [width, height],
    [0, height]
])

matrix = cv2.getPerspectiveTransform(pts_src, pts_dst)

warped = cv2.warpPerspective(img, matrix, (width, height))

warped_rgb = cv2.cvtColor(warped, cv2.COLOR_BGR2RGB)

# grayscale hasil perspektif
warped_gray = cv2.cvtColor(warped, cv2.COLOR_BGR2GRAY)

# blur ringan
blur = cv2.GaussianBlur(warped_gray, (3,3), 0)

# =========================
# 4. ADAPTIVE THRESHOLD
# =========================
adaptive = cv2.adaptiveThreshold(
    blur,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

# =========================
# 5. MENAMPILKAN SEMUA HASIL
# =========================
plt.figure(figsize=(15,10))

# gambar asli
plt.subplot(2,2,1)
plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis("off")

# grayscale
plt.subplot(2,2,2)
plt.imshow(blur, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

# perspektif
plt.subplot(2,2,3)
plt.imshow(warped_rgb)
plt.title("Koreksi Perspektif")
plt.axis("off")

# adaptive threshold
plt.subplot(2,2,4)
plt.imshow(adaptive, cmap='gray')
plt.title("Adaptive Threshold")
plt.axis("off")

plt.tight_layout()
plt.show()
```

>>Program diawali dengan membaca gambar dan mengubahnya menjadi grayscale untuk menyederhanakan informasi citra dengan hanya mempertahankan intensitas cahaya. Setelah itu dilakukan penentuan empat titik sudut papan tulis secara manual untuk proses koreksi perspektif menggunakan transformasi homografi agar objek terlihat lebih lurus dan sejajar. Selanjutnya diterapkan adaptive threshold karena kondisi pencahayaan pada gambar tidak merata akibat pantulan cahaya lampu, sehingga tulisan pada papan tulis dapat terlihat lebih jelas. Seluruh hasil pemrosesan kemudian ditampilkan dalam satu grid agar perubahan kualitas citra pada setiap tahap dapat dibandingkan dengan mudah.
